# BTXRD: 512×512
Makes every image in `train`, `validation` and `test` a **512×512 grayscale PNG**.
The images are replaced in their folders. The full-size images are still in `images` and in `BTXRD dataset`.

```python
from collections import Counter
from pathlib import Path

from PIL import Image

BTXRD = Path("../data/BTXRD no duplicates")
SIZE = (512, 512)
SPLITS = ["train", "validation", "test"]
```

## Before: image sizes

```python
for split in SPLITS:
    sizes = Counter(Image.open(path).size for path in (BTXRD / split).glob("*/*.jp*g"))
    print(split, dict(sizes))
```

## Resize
Images are made smaller with Lanczos, a high-quality resizing method, and saved as PNG, which does not lose quality like JPEG. Grayscale loses nothing, because the red, green and blue values of these X-rays are the same. After the PNG is saved, the JPEG is deleted.

```python
for split in SPLITS:
    for path in sorted((BTXRD / split).glob("*/*.jp*g")):  # */ = normal, tumor
        image = Image.open(path).convert("L")  # L = grayscale
        image = image.resize(SIZE, Image.LANCZOS)
        image.save(path.with_suffix(".png"))
        path.unlink()

print("Done")
```

## After: check the images

```python
for split in SPLITS:
    paths = list((BTXRD / split).glob("*/*.png"))
    sizes = Counter(Image.open(path).size for path in paths)
    modes = Counter(Image.open(path).mode for path in paths)
    jpegs = list((BTXRD / split).glob("*/*.jp*g"))
    print(split, len(paths), "images, sizes:", dict(sizes), "modes:", dict(modes), "JPEGs left:", len(jpegs))
```
