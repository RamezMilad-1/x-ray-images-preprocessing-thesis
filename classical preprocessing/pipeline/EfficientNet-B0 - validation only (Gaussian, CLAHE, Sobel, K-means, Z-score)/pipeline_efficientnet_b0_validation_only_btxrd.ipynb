# Classical pipeline for EfficientNet-B0, chosen on validation only (BTXRD)
Applies 5 classical methods to every X-ray, one after the other, **at its full size** (2440 x 2440), and only then makes it 512 x 512 (the same way as notebook 07).

**Why these methods.** For each step, the method with the best validation accuracy for EfficientNet-B0 (the accuracy on the validation images at the epoch the model kept) is used, but only if it beat the raw images on the validation images. The test images were not used to choose, so the test result of this pipeline stays fair.

1. **Gaussian filter (noise reduction)**: validation accuracy 78.0% (raw images: 77.2%)
2. **CLAHE (contrast enhancement)**: validation accuracy 78.0% (raw images: 77.2%)
3. **Sobel (edge enhancement)**: validation accuracy 79.0% (raw images: 77.2%)
4. **K-means (light ROI extraction)**: validation accuracy 78.0% (raw images: 77.2%)
5. **Z-score normalization (normalization)**: validation accuracy 79.4% (raw images: 77.2%)

Each step works on the result of the step before: X-ray → Gaussian → CLAHE → Sobel → K-means → Z-score → 512 x 512. The settings and the code of each step are the same as in its own notebook, so the only new thing is that the steps are combined. The order is: noise reduction, contrast, edges, the mask, and normalization last. Normalization goes last because it changes the gray values of the whole image (Z-score even turns the black background gray), so the steps after it would take the background for X-ray. The mask comes after the edges, so no edge line is drawn along the border of the mask.

The new images are saved as PNG in **`BTXRD_pipeline_efficientnet_b0_validation_only_512.zip`**, in this folder. Inside the zip are the same `train`, `validation` and `test` folders, each with `normal` and `tumor`, as in `BTXRD_512.zip`. Every image stays in the same split as before.
The original data is only read, never changed.

## Settings

```python
import io
import json
import zipfile
from collections import Counter
from pathlib import Path

import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
from skimage.metrics import peak_signal_noise_ratio, structural_similarity
from scipy import ndimage
from sklearn.cluster import KMeans

DATA = Path("../../../../data/BTXRD no duplicates")    # the original full-size images without duplicates (only read)
ANNOTATIONS = DATA / "Annotations"                     # tumor outlines (only read)
OLD_ZIP = Path("../../../../data/BTXRD_512.zip")       # the baseline zip (only read, to check the split)
NAME = "BTXRD_pipeline_efficientnet_b0_validation_only_512"   # the name of the new zip and of the folder inside it
NEW_ZIP = Path(NAME + ".zip")                          # the new zip is saved here, in this folder
SPLITS = ["train", "validation", "test"]
MALIGNANT = {"osteosarcoma", "other mt"}               # all other tumor types are benign

SIZE = (512, 512)    # the same size as the baseline
# step 1: Gaussian filter (noise reduction)
GAUSSIAN_SIZE = 3      # Gaussian filter: 3 x 3 pixels
GAUSSIAN_SIGMA = 0.8   # how strong the blur is (0.8 is the value OpenCV uses for a 3 x 3 window)
# step 2: CLAHE (contrast enhancement)
CLIP_LIMIT = 2.0     # CLAHE: how strong the contrast boost is
TILES = (8, 8)       # CLAHE: the image is cut into 8 x 8 parts
# step 3: Sobel (edge enhancement)
SIGMA = 5            # Sobel: the blur is 5 pixels wide on the full image (about 1 pixel on the 512 x 512 image)
WEIGHT = 1.0         # Sobel: how much of the edge strength is added to the X-ray
# step 4: K-means (light ROI extraction)
CLUSTERS = 3         # K-means: the gray values of the X-ray are sorted into 3 groups (dark, middle, bright)
CLOSE = 25           # mask clean-up: closes gaps up to 25 pixels wide
OPEN = 25            # mask clean-up: removes specks thinner than 25 pixels
MARGIN = 50          # mask clean-up: 50 pixels extra on every side (about 10 pixels on the 512 x 512 image)
# step 5: Z-score normalization (normalization)
Z_RANGE = 3          # Z-score: z from -3 to +3 is saved as 0 to 255
BLACK = 15           # pixels at or below this are black background
RING = 10            # CII: width of the ring around the tumor, in pixels (on the 512 x 512 image)
```

## List the images
Every image in the `train`, `validation` and `test` folders, with its split, label (normal or tumor) and group (normal, benign or malignant).

```python
def tumor_type(path):
    annotation = ANNOTATIONS / (path.stem + ".json")
    if not annotation.exists():
        return ""                                              # normal images have no annotation
    return json.loads(annotation.read_text())["shapes"][0]["label"]


records = []
for split in SPLITS:
    for path in sorted((DATA / split).glob("*/*.jp*g")):       # */ = normal or tumor, files end in .jpeg or .jpg
        kind = tumor_type(path)
        if kind == "":
            group = "normal"
        elif kind in MALIGNANT:
            group = "malignant"
        else:
            group = "benign"
        records.append({"image": path.stem, "split": split, "label": path.parent.name, "group": group, "path": path})

images = pd.DataFrame(records)
print(len(images), "images")
print(images.groupby(["split", "label"]).size())
```

## Check: the same split as `BTXRD_512.zip`
Every image must be in the same split (train, validation or test) and have the same label as in the baseline zip. If not, the notebook stops here.

```python
with zipfile.ZipFile(OLD_ZIP) as old_zip:
    old_names = {name.removeprefix("BTXRD_512/").removesuffix(".png")   # for example train/normal/IMG001868
                 for name in old_zip.namelist() if name.endswith(".png")}
names = set(images["split"] + "/" + images["label"] + "/" + images["image"])

print("Images in BTXRD_512.zip:", len(old_names), "| images here:", len(names))
print("Same split and label for every image:", names == old_names)
assert names == old_names, "the split is not the same as in BTXRD_512.zip"
```

## Functions
Load an image at full size, the steps (each copied from its own notebook), the pipeline that applies them one after the other, make it 512 x 512, and find the X-ray inside the black square.

```python
def load(path):
    image = Image.open(path).convert("L")                      # gray, pixels 0 (black) to 255 (white)
    return np.array(image)                                     # full size: 2440 x 2440


def gaussian_filter(image):
    size = (GAUSSIAN_SIZE, GAUSSIAN_SIZE)
    return cv2.GaussianBlur(image, size, GAUSSIAN_SIGMA)       # every pixel gets a weighted average of the pixels around it (near pixels count more)


clahe_tool = cv2.createCLAHE(clipLimit=CLIP_LIMIT, tileGridSize=TILES)


def clahe(image):
    return clahe_tool.apply(image)                             # HE in each of the 8 x 8 parts, with a limit


def sobel_edges(image):
    blurred = cv2.GaussianBlur(image, (0, 0), SIGMA)           # a blur first, so the filter finds bone edges, not pixel noise
    gx = cv2.Sobel(blurred, cv2.CV_64F, 1, 0, ksize=3)         # how fast the gray value changes from left to right
    gy = cv2.Sobel(blurred, cv2.CV_64F, 0, 1, ksize=3)         # the same from top to bottom
    strength = np.hypot(gx, gy)                                # edge strength: high on edges, 0 in flat areas
    enhanced = image + WEIGHT * strength                       # the edges get brighter
    return np.clip(enhanced, 0, 255).round().astype(np.uint8)  # keep the gray values between 0 and 255


def disk(size):
    return cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (size, size))   # a round shape, size pixels wide


def clean(mask):
    mask = cv2.morphologyEx(mask.astype(np.uint8), cv2.MORPH_CLOSE, disk(CLOSE))   # close small gaps
    mask = ndimage.binary_fill_holes(mask).astype(np.uint8)   # fill holes, so a dark part inside a bone (like a lytic tumor) stays in
    mask = cv2.morphologyEx(mask, cv2.MORPH_OPEN, disk(OPEN))  # remove small specks
    return cv2.dilate(mask, disk(MARGIN)) == 1                 # a safety margin around the mask; True = keep


def kmeans_roi(image):
    xray = image[image > BLACK]                                # the X-ray pixels, not the black square
    counts = np.bincount(xray, minlength=256)                  # how many pixels have each gray value
    values = np.nonzero(counts)[0]                             # the gray values that occur
    kmeans = KMeans(n_clusters=CLUSTERS, n_init=10, random_state=42)   # random_state: the same result every time
    kmeans.fit(values.reshape(-1, 1).astype(np.float64), sample_weight=counts[values])   # each gray value counts as often as it occurs
    centers = np.sort(kmeans.cluster_centers_.ravel())         # the middle gray value of each group, from dark to bright
    level = (centers[0] + centers[1]) / 2                      # the border between the darkest group and the next one
    mask = clean(image > level)                                # everything except the darkest group (the body), cleaned up
    return np.where(mask, image, 0).astype(np.uint8)           # keep the X-ray inside the mask, black outside


def z_score(image):
    xray = image[image > BLACK]                                # the X-ray pixels, not the black square
    z = (image.astype(np.float64) - xray.mean()) / xray.std()  # 0 = the X-ray's average, 1 = one standard deviation brighter
    scaled = (z + Z_RANGE) / (2 * Z_RANGE) * 255               # -3 -> 0, 0 -> 127.5, +3 -> 255, so it can be saved as a PNG
    return np.clip(scaled, 0, 255).round().astype(np.uint8)    # beyond 3 standard deviations -> 0 or 255


def pipeline(image):
    image = gaussian_filter(image)         # 1. Gaussian filter (noise reduction)
    image = clahe(image)                   # 2. CLAHE (contrast enhancement)
    image = sobel_edges(image)             # 3. Sobel (edge enhancement)
    image = kmeans_roi(image)              # 4. K-means (light ROI extraction)
    image = z_score(image)                 # 5. Z-score normalization (normalization)
    return image


def shrink(image):
    image = Image.fromarray(image).resize(SIZE, Image.LANCZOS)  # 512 x 512, the same way as notebook 07
    return np.array(image)


def xray_box(image):
    rows = np.where((image > BLACK).any(axis=1))[0]            # rows with something brighter than black
    cols = np.where((image > BLACK).any(axis=0))[0]            # the same for columns
    return rows[0], rows[-1] + 1, cols[0], cols[-1] + 1        # top, bottom, left, right
```

## Before and after each step: 2 normal, 2 benign, 2 malignant
All at 512 x 512, the size the models see. Each step is done on the full image, and only the picture is made smaller. Before = the baseline image (the same as in `BTXRD_512.zip`).

```python
picks = pd.concat([images[images["group"] == group].sample(2, random_state=42)
                   for group in ["normal", "benign", "malignant"]])
titles = ['before (baseline)', '1. Gaussian', '2. + CLAHE', '3. + Sobel', '4. + K-means', '5. + Z-score']
steps = [gaussian_filter, clahe, sobel_edges, kmeans_roi, z_score]

fig, axes = plt.subplots(len(picks), len(titles), figsize=(4 * len(titles), 4.4 * len(picks)))
for row, (_, pick) in zip(axes, picks.iterrows()):
    raw = load(pick["path"])
    full = [raw]
    for step in steps:                                         # every step on the full image, on the result of the step before ...
        full.append(step(full[-1]))
    pictures = [shrink(image) for image in full]               # ... then 512 x 512
    assert np.array_equal(pictures[-1], shrink(pipeline(raw))) # the same as the pipeline function
    top, bottom, left, right = xray_box(pictures[0])           # cut away the black square, so the bone is big enough to see
    for ax, picture, title in zip(row, pictures, titles):
        ax.imshow(picture[top:bottom, left:right], cmap="gray", vmin=0, vmax=255)
        ax.set_title(f"{pick['image']} ({pick['group']})\n{title}")
        ax.axis("off")

plt.tight_layout()
plt.show()
```

## Measuring functions
Both images are compared at 512 x 512: the baseline image (before) and the new image (after).
- PSNR and SSIM: how much the image changed (higher = less change). Measured on the X-ray only, not the black square.
- CII: how much more the tumor stands out from the ring around it. Above 1 = it stands out more.

```python
def tumor_mask(name):
    annotation = json.loads((ANNOTATIONS / (name + ".json")).read_text())
    scale = SIZE[0] / annotation["imageWidth"]                 # the outline was drawn on the 2440 x 2440 image
    mask = np.zeros((SIZE[1], SIZE[0]), dtype=np.uint8)
    for shape in annotation["shapes"]:
        if shape["shape_type"] == "polygon":                   # the outline (the rectangles are only boxes)
            points = np.round(np.array(shape["points"]) * scale).astype(np.int32)
            cv2.fillPoly(mask, [points], 1)
    return mask == 1                                           # True inside the tumor


def ring_around(tumor, before):
    kernel = np.ones((2 * RING + 1, 2 * RING + 1), dtype=np.uint8)
    bigger = cv2.dilate(tumor.astype(np.uint8), kernel) == 1   # the tumor made RING pixels bigger
    return bigger & ~tumor & (before > BLACK)                  # the ring: not the tumor, not the black background


def contrast(image, tumor, ring):
    t = image[tumor].mean()                                    # tumor brightness
    r = image[ring].mean()                                     # ring brightness
    return abs(t - r) / (t + r)                                # abs: a tumor can be darker or brighter than the bone


def cii(before, after, tumor, ring):
    if tumor.sum() == 0 or ring.sum() == 0:
        return np.nan                                          # cannot be measured
    contrast_before = contrast(before, tumor, ring)
    if contrast_before == 0:
        return np.nan                                          # cannot divide by 0
    return contrast(after, tumor, ring) / contrast_before


def psnr_ssim(before, after):
    top, bottom, left, right = xray_box(before)                # the X-ray only
    a, b = before[top:bottom, left:right], after[top:bottom, left:right]
    return peak_signal_noise_ratio(a, b, data_range=255), structural_similarity(a, b, data_range=255)
```

## Make every image, save it in the zip, and measure it
For every image: all the steps on the full image, then 512 x 512, then saved as PNG in the zip (`BTXRD_pipeline_efficientnet_b0_validation_only_512/split/label/name.png`).
Then PSNR, SSIM and CII against the baseline image. CII is only for tumor images. This takes some minutes.

```python
rows = []
with zipfile.ZipFile(NEW_ZIP, "w") as new_zip:                 # "w": makes a new zip (an old one with the same name is replaced)
    for n, (_, item) in enumerate(images.iterrows(), start=1):
        raw = load(item["path"])
        before = shrink(raw)                                   # the baseline image
        after = shrink(pipeline(raw))                          # all the steps on the full image, then 512 x 512

        png = io.BytesIO()                                     # make the PNG file in memory ...
        Image.fromarray(after).save(png, format="PNG")
        new_zip.writestr(f"{NAME}/{item['split']}/{item['label']}/{item['image']}.png", png.getvalue())   # ... and put it in the zip

        psnr, ssim = psnr_ssim(before, after)
        row = {"image": item["image"], "label": item["label"], "PSNR": psnr, "SSIM": ssim, "CII": np.nan}
        if item["label"] == "tumor":
            tumor = tumor_mask(item["image"])
            row["CII"] = cii(before, after, tumor, ring_around(tumor, before))
        rows.append(row)

        if n % 500 == 0:
            print(n, "of", len(images), "done")

scores = pd.DataFrame(rows)
print(len(images), "images saved in", NEW_ZIP)
```

## Check the new zip
Counts the images in each folder, checks that all are 512 x 512 gray images, and that every image is in the same split and label as in `BTXRD_512.zip`.

```python
with zipfile.ZipFile(NEW_ZIP) as new_zip:
    files = [name for name in new_zip.namelist() if name.endswith(".png")]
    sizes, modes = Counter(), Counter()
    for name in files:
        image = Image.open(new_zip.open(name))
        sizes[image.size] += 1
        modes[image.mode] += 1                                 # L = gray

for folder, count in sorted(Counter(name.rsplit("/", 1)[0] for name in files).items()):
    print(folder, count)
print("Sizes:", dict(sizes), "| modes:", dict(modes))
new_names = {name.removeprefix(NAME + "/").removesuffix(".png") for name in files}
print("Same images, split and label as BTXRD_512.zip:", new_names == old_names)
print("Zip size:", round(NEW_ZIP.stat().st_size / 1e6), "MB")
```
